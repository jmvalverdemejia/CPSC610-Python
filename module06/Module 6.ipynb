### Sorting Data in Pandas

#### 1. Basic Sorting

```python
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [24, 27, 22, 32],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']}
df = pd.DataFrame(data)

# Sort by Age
sorted_df = df.sort_values(by='Age')
print("Sorted by Age:\n", sorted_df)
```

#### 2. Sorting in Descending Order

```python
# Sort by Age in descending order
sorted_df_desc = df.sort_values(by='Age', ascending=False)
print("Sorted by Age (descending):\n", sorted_df_desc)
```

#### 3. Sorting by Multiple Columns

```python
# Sample DataFrame with duplicate ages
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Edward'],
        'Age': [24, 27, 22, 32, 27],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'New York']}
df = pd.DataFrame(data)

# Sort by Age, then by Name
sorted_df_multi = df.sort_values(by=['Age', 'Name'])
print("Sorted by Age, then by Name:\n", sorted_df_multi)
```

#### 4. Sorting by Index

```python
# Sort by index
sorted_df_index = sorted_df_multi.sort_index()
print("Sorted by index:\n", sorted_df_index)
```

#### 5. Sorting with NaN Values

```python
# Sample DataFrame with NaN values
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [24, None, 22, 32],
        'City': ['New York', 'Los Angeles', 'Chicago', None]}
df = pd.DataFrame(data)

# Sort by Age with NaNs at the beginning
sorted_df_nan = df.sort_values(by='Age', na_position='first')
print("Sorted by Age with NaNs at the beginning:\n", sorted_df_nan)
```

#### 6. Sorting with sort_values on Custom Functions

```python
# Custom sorting function: Sort by string length of the 'Name' column
sorted_df_custom = df.sort_values(by='Name', key=lambda x: x.str.len())
print("Sorted by length of Name:\n", sorted_df_custom)
```

#### 7. Sorting and Resetting Index

```python
# Sort by Age and reset index
sorted_df_reset = df.sort_values(by='Age').reset_index(drop=True)
print("Sorted by Age and reset index:\n", sorted_df_reset)
```

#### 8. In-place Sorting

```python
# In-place sorting by Age
df.sort_values(by='Age', inplace=True)
print("In-place sorted by Age:\n", df)
```

### Filtering Data in Pandas

#### 1. Basic Filtering with Boolean Indexing

```python
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [24, 27, 22, 32],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']}
df = pd.DataFrame(data)

# Filter rows where Age is greater than 25
filtered_df = df[df['Age'] > 25]
print("Filter where Age > 25:\n", filtered_df)
```

#### 2. Filtering with Multiple Conditions

```python
# Filter rows where Age is greater than 25 and City is not 'Chicago'
filtered_df = df[(df['Age'] > 25) & (df['City'] != 'Chicago')]
print("Filter where Age > 25 and City != 'Chicago':\n", filtered_df)

# Filter rows where Age is less than 30 or City is 'New York'
filtered_df = df[(df['Age'] < 30) | (df['City'] == 'New York')]
print("Filter where Age < 30 or City == 'New York':\n", filtered_df)
```

#### 3. Filtering Using isin()

```python
# Filter rows where City is either 'New York' or 'Chicago'
filtered_df = df[df['City'].isin(['New York', 'Chicago'])]
print("Filter where City is in ['New York', 'Chicago']:\n", filtered_df)
```

#### 4. Filtering Using String Methods

```python
# Filter rows where City starts with 'New'
filtered_df = df[df['City'].str.startswith('New')]
print("Filter where City starts with 'New':\n", filtered_df)

# Filter rows where Name contains 'i'
filtered_df = df[df['Name'].str.contains('i')]
print("Filter where Name contains 'i':\n", filtered_df)
```

#### 5. Filtering with query()

```python
# Filter rows where Age is greater than 25
filtered_df = df.query('Age > 25')
print("Query: Age > 25:\n", filtered_df)

# Filter rows where Age is greater than 25 and City is 'Houston'
filtered_df = df.query('Age > 25 and City == "Houston"')
print("Query: Age > 25 and City == 'Houston':\n", filtered_df)
```

#### 6. Filtering with loc[] and iloc[]

```python
# Using loc[] to filter rows where Age is greater than 25
filtered_df = df.loc[df['Age'] > 25]
print("Using loc[] where Age > 25:\n", filtered_df)

# Using iloc[] to filter the first two rows
filtered_df = df.iloc[:2]
print("Using iloc[] for first two rows:\n", filtered_df)
```

#### 7. Filtering with between()

```python
# Filter rows where Age is between 24 and 30
filtered_df = df[df['Age'].between(24, 30)]
print("Filter where Age is between 24 and 30:\n", filtered_df)
```

### Applying Functions in Pandas

#### 1. Applying Functions to Columns or Rows with apply()

```python
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [24, 27, 22, 32],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']}
df = pd.DataFrame(data)

# Apply a lambda function to the 'Age' column to add 1 to each value
df['AgePlusOne'] = df['Age'].apply(lambda x: x + 1)
print("Apply function to 'Age' column:\n", df)

# Apply a lambda function to each row to create a new column 'AgeCategory'
df['AgeCategory'] = df.apply(lambda row: 'Young' if row['Age'] < 25 else 'Adult', axis=1)
print("Apply function to rows:\n", df)
```

#### 2. Applying Element-wise Functions with map()

```python
# Apply a function to the 'City' column to get the length of each city name
df['CityLength'] = df['City'].map(len)
print("Map function to 'City' column:\n", df)

# Apply a dictionary to map values in the 'Name' column to new values
name_map = {'Alice': 'A', 'Bob': 'B', 'Charlie': 'C', 'David': 'D'}
df['NameInitial'] = df['Name'].map(name_map)
print("Map function with dictionary to 'Name' column:\n", df)
```

#### 3. Applying Element-wise Functions to DataFrames with map()

```python
# Sample DataFrame with numerical values
data = {'A': [1, 2, 3], 'B': [4, 5, 6], 'C': [7, 8, 9]}
df_numeric = pd.DataFrame(data)

# Apply a lambda function to square each element in the DataFrame
df_squared = df_numeric.map(lambda x: x ** 2)
print("Map function to square each element:\n", df_squared)
```

#### 4. Applying Functions Using Built-in Methods

```python
# Sample DataFrame with mixed data types
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Score1': [90, 85, 78, 92],
        'Score2': [88, 79, 95, 91]}
df_scores = pd.DataFrame(data)

# Apply sum to each column
column_sums = df_scores[['Score1', 'Score2']].sum()
print("Sum of each column:\n", column_sums)

# Apply mean to each row
row_means = df_scores[['Score1', 'Score2']].mean(axis=1)
print("Mean of each row:\n", row_means)
```

#### 5. Applying Functions with Conditions Using numpy.where()

```python
import numpy as np

# Apply a function to create a new column based on a condition
df_scores['HighScore'] = np.where(df_scores['Score1'] > 80, 'High', 'Low')
print("Apply function conditionally using numpy.where:\n", df_scores)

for index, row in df.iterrows():
	if row['Age'] < 25:
		print(f"{row['Name']} is a young employee from {row['City']}.")
	else:
		print(f"{row['Name']} is an adult employee from {row['City']}.")
```

### Grouping Data

#### 1. Basic Grouping

```python
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Edward'],
        'Age': [24, 27, 22, 32, 27],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'New York'],
        'Score': [85, 92, 78, 95, 88]}
df = pd.DataFrame(data)

# Group by 'City' and calculate the mean age, result is a series
grouped_df = df.groupby('City')['Age'].mean()
print("Mean Age by City:\n", grouped_df)

# Group by 'City' and calculate the mean age, result is a dataframe
grouped_df = df.groupby(['City'])['Age'].mean()
print("Mean Age by City:\n", grouped_df)

# Result is a dataframe
grouped_df = df.groupby('City')['Age'].agg(['mean'])
print("Mean Age by City:\n", grouped_df)
grouped_df.reset_index(inplace=True)
print("Mean Age by City:\n", grouped_df)
```

#### 2. Grouping by Multiple Columns

```python
# Group by 'City' and 'Age', then calculate the mean score
grouped_df = df.groupby(['City', 'Age'])['Score'].mean()
print("Mean Score by City and Age:\n", grouped_df)
```

#### 3. Applying Multiple Aggregation Functions

```python
# Group by 'City' and apply multiple aggregation functions to 'Score'
grouped_df = df.groupby('City')['Score'].agg(['mean', 'min', 'max'])
print("Multiple Aggregations on Score by City:\n", grouped_df)
```

#### 4. Aggregating Multiple Columns

```python
# Group by 'City' and aggregate 'Age' and 'Score' with different functions
grouped_df = df.groupby('City').agg({'Age': 'mean', 'Score': ['min', 'max']})
print("Aggregating Multiple Columns:\n", grouped_df)
```

#### 5. Using Custom Aggregation Functions

```python
# Custom function to calculate range
def range_func(x):
    return x.max() - x.min()

# Group by 'City' and apply custom function to 'Score'
grouped_df = df.groupby('City')['Score'].agg(range_func)
print("Custom Aggregation Function on Score by City (Range):\n", grouped_df)
```

#### 6. Grouping and Filtering

```python
# Group by 'City' and filter groups where mean score is greater than 85
grouped_df = df.groupby('City').filter(lambda x: x['Score'].mean() > 85)
print("Groups where Mean Score > 85:\n", grouped_df)
```

#### 7. Transforming Grouped Data

```python
# Group by 'City' and transform 'Score' to show mean score for each city
df['MeanCityScore'] = df.groupby('City')['Score'].transform('mean')
print("Transform Grouped Data:\n", df)
```

#### 8. Grouping with groupby and apply

```python
# Group by 'City' and apply a custom function
grouped_df = df.groupby('City').apply(lambda x: x.nlargest(1, 'Score'))
print("Group by City and Apply Custom Function (Top Score in Each City):\n", grouped_df)
```

### Merging and Joining DataFrames in Pandas

#### 1. Merging DataFrames

```python
import pandas as pd

# Sample DataFrames
left = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
                     'Age': [24, 27, 22]})
right = pd.DataFrame({'Name': ['Alice', 'Bob', 'David'],
                      'City': ['New York', 'Los Angeles', 'Houston']})

# Merge on 'Name' column
merged_df = pd.merge(left, right, on='Name')
print("Basic Merge on 'Name':\n", merged_df)
```

```python
# Inner join (default)
inner_merge = pd.merge(left, right, on='Name', how='inner')
print("Inner Join:\n", inner_merge)

# Outer join
outer_merge = pd.merge(left, right, on='Name', how='outer')
print("Outer Join:\n", outer_merge)

# Left join
left_merge = pd.merge(left, right, on='Name', how='left')
print("Left Join:\n", left_merge)

# Right join
right_merge = pd.merge(left, right, on='Name', how='right')
print("Right Join:\n", right_merge)
```

```python
# Sample DataFrames
left = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
                     'Age': [24, 27, 22],
                     'City': ['New York', 'Los Angeles', 'Chicago']})
right = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
                      'City': ['New York', 'Los Angeles', 'Houston'],
                      'Score': [85, 92, 78]})

# Merge on 'Name' and 'City' columns
merged_df_multi = pd.merge(left, right, on=['Name', 'City'])
print("Merge on 'Name' and 'City':\n", merged_df_multi)
```

#### 2. Joining DataFrames

```python
# Sample DataFrames
left = pd.DataFrame({'Age': [24, 27, 22]},
                    index=['Alice', 'Bob', 'Charlie'])
right = pd.DataFrame({'City': ['New York', 'Los Angeles', 'Houston']},
                     index=['Alice', 'Bob', 'David'])

# Join on index
joined_df = left.join(right, how='inner')
print("Basic Join on Index:\n", joined_df)
```

```python
# Inner join (default)
inner_join = left.join(right, how='inner')
print("Inner Join:\n", inner_join)

# Outer join
outer_join = left.join(right, how='outer')
print("Outer Join:\n", outer_join)

# Left join
left_join = left.join(right, how='left')
print("Left Join:\n", left_join)

# Right join
right_join = left.join(right, how='right')
print("Right Join:\n", right_join)
```

```python
# Sample DataFrames
df1 = pd.DataFrame({'Name': ['Alice', 'Bob'],
                    'Age': [24, 27]})
df2 = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
                    'City': ['New York', 'Los Angeles', 'Chicago']})

# Set 'Name' as index
df1.set_index('Name', inplace=True)
df2.set_index('Name', inplace=True)

# Join on 'Name' column
joined_df = df1.join(df2)
print("Join on 'Name' column:\n", joined_df)
```

#### 3. Concatenating DataFrames

```python
# Sample DataFrames
df1 = pd.DataFrame({'Name': ['Alice'], 'Age': [24]})
df2 = pd.DataFrame({'Name': ['Bob'], 'Age': [27]})
df3 = pd.DataFrame({'Name': ['Charlie'], 'Age': [22]})

# Concatenate vertically
vertical_concat = pd.concat([df1, df2, df3], axis=0)
print("Concatenate more than two DataFrames Vertically:\n", vertical_concat)
```

### Pivoting and Melting Data in Pandas

#### 1. Pivoting Data

```python
import pandas as pd

# Sample DataFrame
data = {'Date': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-01', '2023-01-02', '2023-01-03'],
        'City': ['New York', 'New York', 'New York', 'Los Angeles', 'Los Angeles', 'Los Angeles'],
        'Temperature': [30, 32, 34, 75, 77, 79]}
df = pd.DataFrame(data)
print(df)

# Pivot data
pivot_df = df.pivot(index='Date', columns='City', values='Temperature')
print("Basic Pivot:\n", pivot_df)
```

```python
# Sample DataFrame
data = {'Date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02'],
        'City': ['New York', 'Los Angeles', 'New York', 'Los Angeles'],
        'Temperature': [30, 75, 32, 77],
        'Humidity': [85, 60, 80, 65]}
df = pd.DataFrame(data)
print(df)

# Pivot table with aggregation
pivot_table_df = df.pivot_table(index='Date', columns='City', values='Temperature', aggfunc='mean')
print("Pivot Table with Aggregation:\n", pivot_table_df)
```

```python
# Pivot table with multiple aggregation functions
pivot_table_multi_df = df.pivot_table(index='Date', columns='City', values=['Temperature', 'Humidity'], aggfunc={'Temperature': 'mean', 'Humidity': 'max'})
print("Pivot Table with Multiple Aggregation Functions:\n", pivot_table_multi_df)
```

#### 2. Melting Data

```python
# Sample DataFrame
data = {'Name': ['Alice', 'Bob'],
        'Math': [85, 90],
        'Science': [88, 92]}
df = pd.DataFrame(data)
print(df)

# Melt data
melted_df = pd.melt(df, id_vars=['Name'], value_vars=['Math', 'Science'], var_name='Subject', value_name='Score')
print("Basic Melt:\n", melted_df)
```

```python
# Sample DataFrame
data = {'Name': ['Alice', 'Bob'],
        'Grade': ['A', 'B'],
        'Math': [85, 90],
        'Science': [88, 92]}
df = pd.DataFrame(data)
print(df)

# Melt data with multiple ID variables
melted_multi_df = pd.melt(df, id_vars=['Name', 'Grade'], value_vars=['Math', 'Science'], var_name='Subject', value_name='Score')
print("Melt with Multiple ID Variables:\n", melted_multi_df)
```

```python
# Sample DataFrame
data = {'Date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02'],
        'City': ['New York', 'Los Angeles', 'New York', 'Los Angeles'],
        'Temperature': [30, 75, 32, 77],
        'Humidity': [85, 60, 80, 65]}
df = pd.DataFrame(data)
print(df)

# Melt data
melted_df = pd.melt(df, id_vars=['Date', 'City'], value_vars=['Temperature', 'Humidity'], var_name='Measure', value_name='Value')

# Aggregate melted data
aggregated_df = melted_df.groupby(['Date', 'City', 'Measure']).mean().reset_index()
print("Melt and Aggregate Data:\n", aggregated_df)
```
