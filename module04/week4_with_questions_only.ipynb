## NumPy Review
### Creating Arrays: From Lists and Built-in Functions

There are many ways to create NumPy arrays:

- From Python lists: `np.array([...])`
- Built-in filled arrays:
  - `np.zeros(shape)`: all zeros
  - `np.ones(shape)`: all ones
  - `np.full(shape, value)`: custom-filled array
- Sequences:
  - `np.arange(start, stop, step)`
  - `np.linspace(start, stop, num_points)`
- Random arrays:
  - `np.random.rand()`, `np.random.randint()`

````python
import numpy as np

arr = np.array([[2, 5, 7, 1],
                [4, 3, 9, 6],
                [8, 2, 1, 5]])

zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
filled = np.full((2, 3), 7)
sequence = np.arange(0, 10, 2)
linspace = np.linspace(0, 1, 5)
random_ints = np.random.randint(1, 10, size=(3, 4))

print("From list:\n", arr)
print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Filled with 7:\n", filled)
print("Arange:\n", sequence)
print("Linspace:\n", linspace)
print("Random ints:\n", random_ints)
````

### Viewing Array Attributes

To view basic properties of arrays:
- `.shape`: the dimensions of the array
- `.size`: total number of elements
- `.dtype`: data type of elements

````python
print("Array:\n", arr)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Datatype:", arr.dtype)
````

### Indexing and Slicing

We can access parts of a NumPy array using:
- Indexing: `arr[0]` gets the first row
- Column access: `arr[:, 1]` gets the second column
- Boolean indexing: `arr[arr > 5]` selects elements greater than 5

````python
print("First row:", arr[0])
print("Second column:", arr[:, 1])

print("Elements > 5:\n", arr[arr > 5])
````

### Mathematical Operations (ufuncs)

NumPy supports vectorized operations on arrays:
- `np.square()`: squares each element
- Division and multiplication apply to all elements
- `np.log1p()`: computes log(1 + x), safer for small values

````python
print("Original array:\n", arr)
print("Squared:\n", np.square(arr))
print("Half values:\n", arr / 2)
print("Logarithm (log1p):\n", np.log1p(arr))
````

### Statistical Functions

NumPy provides tools to describe data:
- `np.mean()`: average value
- `np.median()`: middle value
- `np.std()`, `np.var()`: measure of spread

````python
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Variance:", np.var(arr))
````

### Summation and Cumulative Functions

These help track totals or running totals:
- `np.sum()`: total sum
- `np.sum(..., axis=1)`: sum of each row
- `np.cumsum()`: running total
- `np.prod()`: product of all elements

````python
print("Sum of all elements:", np.sum(arr))
print("Sum along rows:", np.sum(arr, axis=1))
print("Sum along columns:", np.sum(arr, axis=0))
print("Cumulative sum (flattened):", np.cumsum(arr))
print("Product of all elements:", np.prod(arr))
````

### Comparison and Boolean Masks

You can compare array values and extract positions using:
- `>` and `<` for thresholding
- `==` for exact matches
- `np.where()` to find matching indices

````python
print("Array:\n", arr)
print("Greater than 6:\n", arr > 6)
print("Equal to 3:\n", arr == 3)
print("Where values > 5:", np.where(arr > 5))
````

### Searching and Sorting

- `np.where(condition)`: returns indices where condition is True
- `np.argmax()`, `np.argmin()`: index of max/min value
- `np.sort(array)`: returns a sorted copy (does not change original)
- Sorting along rows or columns: use `axis` parameter

````python
print("Original array:\n", arr)
print("Where values > 6:", np.where(arr > 6))
print("Index of max value:", np.argmax(arr))
print("Index of min value:", np.argmin(arr))
print("Sorted (flattened):", np.sort(arr, axis=None))
print("Sorted rows:\n", np.sort(arr, axis=1))
print("Sorted rows:\n", np.sort(arr, axis=0))
````

## Part 2: Visualizing Arrays as Images

NumPy arrays can be displayed visually as **images**. This is a great way to **see numerical patterns** or **represent data visually**.

This actually works **both ways**:

- You can **visualize arrays** as images
- But also: **images loaded for machine learning tasks are actually just loaded as NumPy arrays**


In this example, we generate a **100×100 array** of random values between 0 and 1.

To display it as an image, we use:

```python
plt.imshow(array, cmap='gray')
```

`plt.imshow(...)`: Displays a 2D array as an image
- Each number in the array becomes a **pixel**
- By default, higher numbers are brighter
- `cmap='gray'`:
  - Maps values from **0.0 (black)** to **1.0 (white)**
  - This creates a grayscale image

````python
noise = np.random.rand(100, 100)

plt.imshow(noise, cmap='gray')
plt.axis('off')
plt.show()
````
