### Subplots

```python
plt.subplot(2, 1, 1)  # Creating the first subplot in a 2x1 grid
plt.plot([1, 2, 3, 4], [10, 20, 25, 30])  # Plotting the first set of data
plt.title("First Subplot")  # Adding a title to the first subplot

plt.subplot(2, 1, 2)  # Creating the second subplot in a 2x1 grid
plt.plot([1, 2, 3, 4], [30, 25, 20, 10])  # Plotting the second set of data
plt.title("Second Subplot")  # Adding a title to the second subplot

plt.tight_layout()  # Adjusts spacing between subplots to prevent overlap
plt.show()  # Displaying the subplots
```

```python
import matplotlib.gridspec as gridspec  # Importing the gridspec module

# Create a grid of 2x2
gs = gridspec.GridSpec(2, 2)

# Create a larger subplot on the left
plt.subplot(gs[0, 0:2])  # Occupying the top row across two columns
plt.plot([1, 2, 3, 4], [10, 20, 25, 30])
plt.title("Large Subplot")  # Adding a title

# Create a smaller subplot on the bottom-left
plt.subplot(gs[1, 0])  # Occupying the bottom left cell
plt.plot([1, 2, 3, 4], [30, 25, 20, 10])
plt.title("Small Subplot 1")  # Adding a title

# Create another smaller subplot on the bottom-right
plt.subplot(gs[1, 1])  # Occupying the bottom right cell
plt.plot([1, 2, 3, 4], [15, 18, 25, 30])
plt.title("Small Subplot 2")  # Adding a title

plt.tight_layout()  # Adjusting layout to prevent overlap
plt.show()  # Displaying the subplots
```

### Multiple Plots on the Same Axes

```python
import numpy as np  # Importing NumPy for creating data arrays
x = np.linspace(0, 2 * np.pi, 100)  # Creating X-axis data from 0 to 2*pi
y1 = np.sin(x)  # Calculating Y1 data (sin of x)
y2 = np.cos(x)  # Calculating Y2 data (cos of x)

plt.plot(x, y1, label='sin(x)')  # Plotting sin(x) with a label
plt.plot(x, y2, label='cos(x)')  # Plotting cos(x) with a label
plt.title("Sin and Cos Waves")  # Adding a title
plt.xlabel("X-axis")  # Labeling the X-axis
plt.ylabel("Y-axis")  # Labeling the Y-axis
plt.legend()  # Adding a legend to differentiate the waves
plt.show()  # Displaying the plot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## sample data

```python
# Creating the Tips dataset
np.random.seed(42)
n_samples = 200
days = ['Thur', 'Fri', 'Sat', 'Sun']
times = ['Lunch', 'Dinner']
genders = ['Male', 'Female']

tips = pd.DataFrame({
    'total_bill': np.round(np.random.uniform(5, 50, n_samples), 2),
    'tip': np.round(np.random.uniform(1, 10, n_samples), 2),
    'sex': np.random.choice(genders, n_samples),
    'smoker': np.random.choice(['Yes', 'No'], n_samples),
    'day': np.random.choice(days, n_samples),
    'time': np.random.choice(times, n_samples),
    'size': np.random.randint(1, 6, n_samples)
})
```

```python
numeric_cols = tips.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_cols.corr()
corr_matrix
```

```python
# Creating custom data for regression analysis
np.random.seed(42)
advertising_data = pd.DataFrame({
    'TV': np.random.uniform(50, 300, n_samples),
    'Radio': np.random.uniform(20, 100, n_samples),
    'Newspaper': np.random.uniform(10, 80, n_samples),
    'Sales': np.random.uniform(5, 25, n_samples)
})
advertising_data['Sales'] += (0.05 * advertising_data['TV'] + 
                              0.1 * advertising_data['Radio'] +
                              0.02 * advertising_data['Newspaper'] +
                              np.random.normal(0, 2, n_samples))
```

## Basic Plotting Functions in Seaborn

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()
```

### 2.	Distribution Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Example: Histogram of 'total_bill'
sns.histplot(tips['total_bill'], bins=20, color='blue', kde=False)
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# Example: KDE plot of 'total_bill'
sns.kdeplot(tips['total_bill'], color='green', lw=2, fill=True)
plt.title("KDE Plot of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.show()
```

```python
# Create the boxplot
sns.boxplot(x='sex', y='tip', data=tips)

# Add title and labels
plt.title('Boxplot of Tips by Sex')
plt.xlabel('Sex')
plt.ylabel('Tip Amount')

# Show the plot
plt.show()
```

```python
# Example: Violin plot of 'total_bill'
sns.violinplot(y=tips['total_bill'], color='purple', inner="quartile")
plt.title("Violin Plot of Total Bill")
plt.xlabel("Total Bill")
plt.show()
```

```python
# Example: Violin plot of total bill by day
sns.violinplot(x='day', y='total_bill', data=tips, palette='coolwarm')
plt.title("Violin Plot of Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

### 4.	Regression Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
tips = sns.load_dataset("tips")

# Example: Regression plot of 'total_bill' vs 'tip'
sns.regplot(x='total_bill', y='tip', data=tips, scatter_kws={'s':20}, line_kws={'color':'red'})
plt.title("Regression Plot of Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Example: lmplot of 'total_bill' vs 'tip', categorized by 'smoker'
sns.lmplot(x='total_bill', y='tip', hue='smoker', data=tips, markers=['o', 'x'], palette='Set1')
plt.title("LM Plot of Total Bill vs Tip by Smoker")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

## Customizing Your Visuals

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from matplotlib.dates import DateFormatter

# Create datetime-indexed data
dates = pd.date_range("2024-01-01", periods=10, freq="D")
np.random.seed(42)
visitors = np.random.randint(80, 150, size=10)
sales = visitors * 0.6 + np.random.normal(0, 5, size=10)

df = pd.DataFrame({"date": dates, "visitors": visitors, "sales": sales})
df.set_index("date", inplace=True)

# Choose palette colours
seq_color = sns.color_palette("Blues", 6)[4]
qual_color = sns.color_palette("Set1")[0]

# Create figure and axis
fig, ax = plt.subplots(figsize=(14, 6))

# Plot lines
ax.plot(df.index, df["visitors"], label="Daily Visitors", color=seq_color,
        marker="o", linestyle="--", linewidth=2)
ax.plot(df.index, df["sales"], label="Daily Sales", color=qual_color,
        marker="s", linestyle="-", linewidth=2)

# Format dates
ax.xaxis.set_major_formatter(DateFormatter("%b %d"))
plt.xticks(rotation=45)

# Annotation
highlight_date = df.index[5]
highlight_value = df["visitors"].iloc[5]
ax.annotate("Notable point", xy=(highlight_date, highlight_value),
            xytext=(highlight_date, highlight_value + 15),
            arrowprops={"arrowstyle": "->", "color": "gray"})

# Labels and layout
ax.set_title("Daily Visitors and Sales Over Time")
ax.set_xlabel("Date")
ax.set_ylabel("Value")
ax.legend()
plt.tight_layout()
plt.show()
```
